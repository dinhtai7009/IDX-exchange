# Decision Tree Regressor

```python
#initialize decision tree regressor
tree = DecisionTreeRegressor(random_state=30, max_depth = 30, min_samples_split = 200)
```

```python
tree.fit(X_train, Y_train)
preds = tree.predict(X_test)
print(r2_score(Y_test, preds))
print(root_mean_squared_error(Y_test, preds))
```

```python
plt.scatter(preds, Y_test, color='blue')
plt.plot([0, max(preds)], [0, max(preds)], color = 'black')
plt.xlabel('predictions')
plt.ylabel('actual')
plt.show()
```

# Random Forest Regressor

```python
#define random forest regressor
rf = RandomForestRegressor(random_state=30, max_depth = 30, min_samples_split = 200)
rf.fit(X_train, Y_train)
rf_preds = rf.predict(X_test)
print(r2_score(Y_test, rf_preds))
print(root_mean_squared_error(Y_test, rf_preds))
```

```python
#plotting predictions vs. actual
plt.scatter(rf_preds, Y_test, color='blue')
plt.plot([0, max(rf_preds)], [0, max(rf_preds)], color = 'black')
plt.xlabel('predictions')
plt.ylabel('actual')
plt.show()
```
